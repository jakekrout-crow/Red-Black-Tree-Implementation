```python
class Node:
    
    RED = 'red'
    BLACK = 'black'
    
    def __init__(self, key, value=None):
        self.key = key
        self.value = value
        self.color = self.RED   
        self.left = None
        self.right = None
        self.parent = None

    def __repr__(self):
        return f"Node({self.key}, color={self.color})"


class RedBlackTree:
    
    def __init__(self):
        self.NIL = Node(None) 
        self.NIL.color = self.NIL.BLACK
        self.root = self.NIL
        self.size = 0
      
    def left_rotate(self, x):
        if x is self.NIL or x.right is self.NIL:
            return
        
        y = x.right
        x.right = y.left
        
        if y.left is not self.NIL:
            y.left.parent = x
        
        y.parent = x.parent
        
        if x.parent is self.NIL:
            self.root = y
        elif x is x.parent.left:
            x.parent.left = y
        else:
            x.parent.right = y
        
        y.left = x
        x.parent = y
    
    def right_rotate(self, x):
        if x is self.NIL or x.left is self.NIL:
            return
        
        y = x.left
        x.left = y.right
        
        if y.right is not self.NIL:
            y.right.parent = x
        
        y.parent = x.parent
        
        if x.parent is self.NIL:
            self.root = y
        elif x is x.parent.right:
            x.parent.right = y
        else:
            x.parent.left = y
        
        y.right = x
        x.parent = y
     
    def flip_colors(self, node):
        if node is self.NIL:
            return
        
        node.color = self.NIL.BLACK if node.color == self.NIL.BLACK else self.NIL.RED
        if node.left is not self.NIL:
            node.left.color = self.NIL.BLACK if node.left.color == self.NIL.BLACK else self.NIL.RED
        if node.right is not self.NIL:
            node.right.color = self.NIL.BLACK if node.right.color == self.NIL.BLACK else self.NIL.RED
        
    def insert(self, key, value=None):
        new_node = Node(key, value)
        new_node.left = self.NIL
        new_node.right = self.NIL
        
        parent = self.NIL
        current = self.root
        
        while current is not self.NIL:
            parent = current
            if key < current.key:
                current = current.left
            elif key > current.key:
                current = current.right
            else:
                current.value = value
                return
        
        new_node.parent = parent
        
        if parent is self.NIL:
            self.root = new_node
        elif key < parent.key:
            parent.left = new_node
        else:
            parent.right = new_node
        
        self._insert_fixup(new_node)
        self.size += 1
    
    def _insert_fixup(self, z):
        while z.parent.color == self.NIL.RED:
            if z.parent is z.parent.parent.left:
                uncle = z.parent.parent.right
                
                if uncle.color == self.NIL.RED:
                    z.parent.color = self.NIL.BLACK
                    uncle.color = self.NIL.BLACK
                    z.parent.parent.color = self.NIL.RED
                    z = z.parent.parent
                
                elif z is z.parent.right:
                    z = z.parent
                    self.left_rotate(z)
                
                else:
                    z.parent.color = self.NIL.BLACK
                    z.parent.parent.color = self.NIL.RED
                    self.right_rotate(z.parent.parent)
            else:
                uncle = z.parent.parent.left
                
                if uncle.color == self.NIL.RED:
                    z.parent.color = self.NIL.BLACK
                    uncle.color = self.NIL.BLACK
                    z.parent.parent.color = self.NIL.RED
                    z = z.parent.parent
                
                elif z is z.parent.left:
                    z = z.parent
                    self.right_rotate(z)
                
                else:
                    z.parent.color = self.NIL.BLACK
                    z.parent.parent.color = self.NIL.RED
                    self.left_rotate(z.parent.parent)
        
        self.root.color = self.NIL.BLACK 
    
    def delete(self, key):
        z = self.search_node(key)
        if z is self.NIL:
            raise KeyError(f"The key {key} was not found in this tree.")
        
        self._delete(z)
        self.size -= 1
    
    def _delete(self, z):
        y = z
        y_original_color = y.color
        x = None
        
        if z.left is self.NIL:
            x = z.right
            self._transplant(z, z.right)
        elif z.right is self.NIL:
            x = z.left
            self._transplant(z, z.left)
        else:
            y = self._minimum(z.right)
            y_original_color = y.color
            x = y.right
            
            if y.parent is z:
                x.parent = y
            else:
                self._transplant(y, y.right)
                y.right = z.right
                y.right.parent = y
            
            self._transplant(z, y)
            y.left = z.left
            y.left.parent = y
            y.color = z.color
        
        if y_original_color == self.NIL.BLACK:
            self._delete_fixup(x)
    
    def _delete_fixup(self, x):
        while x is not self.root and x.color == self.NIL.BLACK:
            if x is x.parent.left:
                w = x.parent.right
                
                if w.color == self.NIL.RED:
                    w.color = self.NIL.BLACK
                    x.parent.color = self.NIL.RED
                    self.left_rotate(x.parent)
                    w = x.parent.right
                
                if (w.left.color == self.NIL.BLACK and 
                    w.right.color == self.NIL.BLACK):
                    w.color = self.NIL.RED
                    x = x.parent
                else:
                    if w.right.color == self.NIL.BLACK:
                        w.left.color = self.NIL.BLACK
                        w.color = self.NIL.RED
                        self.right_rotate(w)
                        w = x.parent.right
                    
                    w.color = x.parent.color
                    x.parent.color = self.NIL.BLACK
                    w.right.color = self.NIL.BLACK
                    self.left_rotate(x.parent)
                    x = self.root
            else:
                w = x.parent.left
                
                if w.color == self.NIL.RED:
                    w.color = self.NIL.BLACK
                    x.parent.color = self.NIL.RED
                    self.right_rotate(x.parent)
                    w = x.parent.left
                
                if (w.right.color == self.NIL.BLACK and 
                    w.left.color == self.NIL.BLACK):
                    w.color = self.NIL.RED
                    x = x.parent
                else:
                    if w.left.color == self.NIL.BLACK:
                        w.right.color = self.NIL.BLACK
                        w.color = self.NIL.RED
                        self.left_rotate(w)
                        w = x.parent.left
                    
                    w.color = x.parent.color
                    x.parent.color = self.NIL.BLACK
                    w.left.color = self.NIL.BLACK
                    self.right_rotate(x.parent)
                    x = self.root
        
        x.color = self.NIL.BLACK
    
    def _transplant(self, u, v):
        if u.parent is self.NIL:
            self.root = v
        elif u is u.parent.left:
            u.parent.left = v
        else:
            u.parent.right = v
        v.parent = u.parent
    
    def search(self, key):
        node = self.search_node(key)
        if node is not self.NIL:
            return node
        return None
    
    def search_node(self, key):
        current = self.root
        while current is not self.NIL:
            if key == current.key:
                return current
            elif key < current.key:
                current = current.left
            else:
                current = current.right
        return self.NIL
    
    def in_order(self):
        result = []
        self._in_order_helper(self.root, result)
        return result
    
    def _in_order_helper(self, node, result):
        if node is not self.NIL:
            self._in_order_helper(node.left, result)
            result.append(node.key)
            self._in_order_helper(node.right, result)
    
    def _minimum(self, node):
        current = node
        while current is not self.NIL and current.left is not self.NIL:
            current = current.left
        return current
    
    def _maximum(self, node):
        current = node
        while current is not self.NIL and current.right is not self.NIL:
            current = current.right
        return current
    
    def successor(self, node):
        if node.right is not self.NIL:
            return self._minimum(node.right)
        parent = node.parent
        while parent is not self.NIL and node is parent.right:
            node = parent
            parent = parent.parent
        return parent
    
    def predecessor(self, node):
        if node.left is not self.NIL:
            return self._maximum(node.left)
        parent = node.parent
        while parent is not self.NIL and node is parent.left:
            node = parent
            parent = parent.parent
        return parent
    
    def height(self):
        return self._height_helper(self.root)
    
    def _height_helper(self, node):
        if node is self.NIL:
            return 0
        return 1 + max(self._height_helper(node.left), 
                       self._height_helper(node.right))
    
    def is_empty(self):
        return self.root is self.NIL
    

    def verify_rb_properties(self):
        errors = []
        
        if self.root is not self.NIL and self.root.color == self.NIL.RED:
            errors.append("Property 2 violated: The root node is red (must be black).")

        red_violations = self._check_no_consecutive_reds(self.root)
        if red_violations:
            for key in red_violations:
                errors.append(
                    f"Property 4 violated: The node with the key {key} is red "
                    f"but has a red child."
                )
        
        black_height_violations = self._check_black_height_consistency(self.root)
        if black_height_violations:
            for info in black_height_violations:
                errors.append(
                    f"Property 5 violated: {info}"
                )
        
        is_valid = len(errors) == 0
        return is_valid, errors
    
    def _check_no_consecutive_reds(self, node):
        violations = []
        if node is self.NIL:
            return violations
        
        if node.color == self.NIL.RED:
            if (node.left is not self.NIL and node.left.color == self.NIL.RED):
                violations.append(node.key)
            if (node.right is not self.NIL and node.right.color == self.NIL.RED):
                violations.append(node.key)
        
        violations.extend(self._check_no_consecutive_reds(node.left))
        violations.extend(self._check_no_consecutive_reds(node.right))
        
        return violations
    
    def _check_black_height_consistency(self, node):
        if node is self.NIL:
            return []
        left_bh = self._compute_black_height(node.left)
        right_bh = self._compute_black_height(node.right)
        
        violations = []
        if left_bh != right_bh:
            violations.append(
                f"The path through node {node.key} has a black height of: "
                f"left={left_bh}, right={right_bh}."
            )
        
        violations.extend(self._check_black_height_consistency(node.left))
        violations.extend(self._check_black_height_consistency(node.right))
        
        return violations
    
    def _compute_black_height(self, node):
        if node is self.NIL:
            return 0
        
        bh = self._compute_black_height(node.left)
        if node.color == self.NIL.BLACK:
            bh += 1
        
        return bh
    
    def __str__(self):
        return str(self.in_order())
    
    def __len__(self):
        return self.size
    
    def __contains__(self, key):
        return self.search(key) is not None
```
